```python
# 1. input.txt файлыг үүсгэх
tests = """Уртнасан Хүрэлбаатар Эмэгтэй БС АХС 96864108
Эрхбаяр Тогтох Эмэгтэй МБУС Т2 90889308
Бадамсүрэн Ренчин Эмэгтэй Лицей ГаТ 86223186
Гантогтох Бор Эрэгтэй АХБС Т2 80404926
Уламбаяр Ганбаяр Эмэгтэй Лицей Т3 86484938
Гармаа Сэсээр Эрэгтэй БС ОХТ 91539913
Баттогтох Сосор Эрэгтэй БТС Т3 99470642
Гоомарал Сүрэн Эрэгтэй БоСС Т4 95832190
Золзаяа Сэрээнэн Эмэгтэй ДУТС БиоТ 80634254
Эрхбаяр Батбаатар Эмэгтэй НХУС Т6 95674121
"""

with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(tests)

# 2. Бодлогын код (output.txt файлыг үүсгэнэ)
ifile = open('input.txt', 'r', encoding='utf-8')
mylines = ifile.readlines()
ifile.close()

companies = {
    'Mobicom': ['94', '95', '99', '85'],
    'Unitel':  ['80', '86', '88', '89'],
    'Skytel':  ['90', '91', '96'],
    'Gmobile': ['83', '93', '97'],
}

count = {}
for line in mylines:
    values = line.split()
    if len(values) < 6:
        continue
    prefix = values[5][:2]
    for name, prefixes in companies.items():
        if prefix in prefixes:
            if name in count:
                count[name] += 1
            else:
                count[name] = 1

maxCount = max(count.values())

ofile = open('output.txt', 'w', encoding='utf-8')
for name in sorted(count):
    if count[name] == maxCount:
        ofile.write(name + ' ' + str(count[name]) + '\n')
ofile.close()

# 3. Үр дүнг хэвлэж шалгах
print("--- output.txt файлын агуулга ---")
print(open('output.txt', encoding='utf-8').read())
```

```python
#Бодлого 6_02
# input.txt файлаас мөрүүдийг унших
ifile = open('input.txt', 'r', encoding='utf-8')
mylines = ifile.readlines()
ifile.close()

# Үүрэн холбооны операторуудын угтвар цифрүүд
companies = {
    'Мобиком':  ['99', '95', '94', '85'],
    'Юнител':   ['88', '89', '80', '86'],
    'Скайтел':  ['91', '90', '96'],
    'Жимобайл': ['93', '97', '83'],
}

count = {}
for line in mylines:
    values = line.split()
    if len(values) < 6:          # Хоосон мөрийг алгасах
        continue
    prefix = values[5][:2]       # Утасны дугаарын эхний 2 цифр

    for name, prefixes in companies.items():
        if prefix in prefixes:
            if name in count:
                count[name] += 1
            else:
                count[name] = 1

# Хамгийн их давтамжтай (олон) хүний тоог олох
maxCount = max(count.values())

# output.txt файлд гаралтын хэлбэрийн дагуу бичих
ofile = open('output.txt', 'w', encoding='utf-8')
for name in sorted(count):
    if count[name] == maxCount:
        ofile.write(name + ' ' + str(count[name]) + '\n')
ofile.close()

# Үр дүнг дэлгэц дээр шалгаж харуулах
print(open('output.txt', encoding='utf-8').read())
```

```python
# 6.2
def is_fibonacci_and_index(n):
    if n < 0:
        return False, -1

    a, b = 0, 1
    index = 1

    if n == 0:
        return True, 0

    while b < n:
        a, b = b, a + b
        index += 1

    if b == n:
        return True, index
    else:
        return False, -1

# Жишээгээр шалгах
test_numbers = [5, 8, 21, 4, 13]

for num in test_numbers:
    is_fib, idx = is_fibonacci_and_index(num)
    if is_fib:
        print(f"{num} нь Фибоначийн тоо мөн. Дарааллын {idx}-р гишүүн байна.")
    else:
        print(f"{num} нь Фибоначийн тоо биш байна.")
```
