```python
# 6.2
def is_fibonacci_and_index(n):
    if n < 0:
        return False, -1

    a, b = 0, 1
    index = 1

    if n == 0:
        return True, 0

    while b < n:
        a, b = b, a + b
        index += 1

    if b == n:
        return True, index
    else:
        return False, -1

# Жишээгээр шалгах
test_numbers = [5, 8, 21, 4, 13]

for num in test_numbers:
    is_fib, idx = is_fibonacci_and_index(num)
    if is_fib:
        print(f"{num} нь Фибоначийн тоо мөн. Дарааллын {idx}-р гишүүн байна.")
    else:
        print(f"{num} нь Фибоначийн тоо биш байна.")
```

```python
#6.03
class Rectangle:
    # Байгуулагч функц (Өгөгдлийг ямар хэлбэрээр өгснөөс хамаарч утга онооно)
    def __init__(self, *args):
        # 4 хувилбараар өгөгдөл авах нөхцөлүүд:
        if len(args) == 0:
            # 1. Гараас унших хувилбар
            self.name = input("Тэгш өнцөгтийн нэрийг оруулна уу: ")
            self.a = float(input("Урт (a)-г оруулна уу: "))
            self.b = float(input("Өргөн (b)-г оруулна уу: "))

        elif len(args) == 3 and isinstance(args[0], str):
            # 2. Параметрээр шууд утга авах хувилбар: Rectangle("Rec1", 5, 3)
            self.name = args[0]
            self.a = args[1]
            self.b = args[2]

        elif len(args) == 1 and isinstance(args[0], Rectangle):
            # 3. Өөр объектоос хуулж авах хувилбар: Rectangle(other_rect)
            other = args[0]
            self.name = other.name + "_copy"
            self.a = other.a
            self.b = other.b

        elif len(args) == 1 and isinstance(args[0], str):
            # 4. Файлаас унших хувилбар (Файлын нэрийг дамжуулсан үед)
            # Жишээ нь файлын эхний мөрөнд: ТэгшӨнцөгт1 10 4 гэж зайгаар тусгаарлагдсан байгаа гэж үзнэ.
            filename = args[0]
            try:
                with open(filename, 'r', encoding='utf-8') as f:
                    line = f.readline().split()
                    if len(line) >= 3:
                        self.name = line[0]
                        self.a = float(line[1])
                        self.b = float(line[2])
                    else:
                        self.name, self.a, self.b = "FileRect", 1.0, 1.0
            except FileNotFoundError:
                self.name, self.a, self.b = "FileRect", 1.0, 1.0

    # Талбай тооцох гишүүн функц
    def area(self):
        return self.a * self.b

    # Периметр тооцох гишүүн функц
    def perimeter(self):
        return 2 * (self.a + self.b)

    # Мэдээллийг хэвлэх туслах функц
    def display(self):
        print(f"Нэр: {self.name}, Урт (a): {self.a}, Өргөн (b): {self.b}, Талбай: {self.area()}, Периметр: {self.perimeter()}")


# --- Дээрх 4 хувилбараар 4 объект үүсгэж ашиглах нь ---

print("--- 1. Параметрээр авсан объект ---")
r1 = Rectangle("Нэгдүгээр", 10, 5)
r1.display()

print("\n--- 2. Өөр объектоос хуулж авсан объект ---")
r2 = Rectangle(r1)
r2.display()

print("\n--- 3. Файлаас уншиж үүсгэх объект ---")
# Эхлээд туршилтын файл үүсгэх
with open("rect_data.txt", "w", encoding='utf-8') as f:
    f.write("Гуравдугаар_Тэгшөнцөгт 8 4\n")

r3 = Rectangle("rect_data.txt")
r3.display()

print("\n--- 4. Гараас уншиж үүсгэх объект ---")
# (Энэ хэсэг ажиллахад танаас гараас утга оруулахыг хүснэ)
r4 = Rectangle()
r4.display()
```

```python
import math

class EqQuad:
    # Байгуулагч функц (4 өөр хувилбараар өгөгдөл авах боломжтой)
    def __init__(self, *args):
        if len(args) == 0:
            # 1. Гараас унших хувилбар
            print("Гараас квадрат тэгшитгэлийн коэфициентүүдийг оруулна уу:")
            self.a = float(input("a = "))
            self.b = float(input("b = "))
            self.c = float(input("c = "))

        elif len(args) == 3:
            # 2. Параметрээр шууд утга авах хувилбар: EqQuad(1, -3, 2)
            self.a = args[0]
            self.b = args[1]
            self.c = args[2]

        elif len(args) == 1 and isinstance(args[0], EqQuad):
            # 3. Өөр объектоос хуулж авах хувилбар: EqQuad(other_eq)
            other = args[0]
            self.a = other.a
            self.b = other.b
            self.c = other.c

        elif len(args) == 1 and isinstance(args[0], str):
            # 4. Файлаас унших хувилбар (Файлын нэрийг дамжуулах үед)
            # Файлын эхний мөрөнд a, b, c утгууд зайгаар тусгаарлагдан бичигдсэн байна.
            filename = args[0]
            try:
                with open(filename, 'r', encoding='utf-8') as f:
                    line = f.readline().split()
                    if len(line) >= 3:
                        self.a = float(line[0])
                        self.b = float(line[1])
                        self.c = float(line[2])
                    else:
                        self.a, self.b, self.c = 1.0, -5.0, 6.0
            except FileNotFoundError:
                self.a, self.b, self.c = 1.0, -5.0, 6.0

    # Тэгшитгэлийн шийд олох гишүүн функц
    def bodolt(self):
        print(f"\nТэгшитгэл: {self.a}x^2 + {self.b}x + {self.c} = 0")

        if self.a == 0:
            print("Энэ нь квадрат тэгшитгэл биш байна (a = 0).")
            return

        # Дискриминант олох (D = b^2 - 4ac)
        d = self.b**2 - 4 * self.a * self.c
        print(f"Дискриминант (D) = {d}")

        if d > 0:
            x1 = (-self.b + math.sqrt(d)) / (2 * self.a)
            x2 = (-self.b - math.sqrt(d)) / (2 * self.a)
            print(f"Тэгшитгэл хоёр язгууртай байна: x1 = {x1}, x2 = {x2}")
        elif d == 0:
            x = -self.b / (2 * self.a)
            print(f"Тэгшитгэл нэг давхар язгууртай байна: x = {x}")
        else:
            print("Тэгшитгэл бодит язгуургүй байна (D < 0).")


# --- Дээрх 4 хувилбараар 4 объект үүсгэж бодолтыг хийх нь ---

print("--- 1. Параметрээр өгсөн объект ---")
eq1 = EqQuad(1, -3, 2)
eq1.bodolt()

print("\n--- 2. Өөр объектоос хуулж авсан объект ---")
eq2 = EqQuad(eq1)
eq2.bodolt()

print("\n--- 3. Файлаас уншиж үүсгэх объект ---")
# Туршилтын файл үүсгэх (a=1, b=-5, c=6)
with open("eq_data.txt", "w", encoding='utf-8') as f:
    f.write("1 -5 6\n")

eq3 = EqQuad("eq_data.txt")
eq3.bodolt()

print("\n--- 4. Гараас уншиж үүсгэх объект ---")
# (Энэ хэсэг ажиллахад танаас a, b, c утгуудыг гараас оруулахыг хүснэ)
eq4 = EqQuad()
eq4.bodolt()
```
