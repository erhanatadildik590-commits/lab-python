```python
#Loop01.
# Өгөгдлийг зайгаар салгаж жагсаалт болгон унших
ugugdul = list(map(int, input().split()))

N = ugugdul[0]  # Эхний тоо - гишүүдийн тоо
toonuud = ugugdul[1:]  # Үлдсэн тоонуудын жагсаалт

niilber = 0
i = 0

# While давталт ашиглан нийлбэр олох
while i < len(toonuud):
  niilber = niilber + toonuud[i]
  i = i + 1

print(niilber)
```

```python
#loop2
import sys

# Бүх оролтыг уншиж зайгаар салгах
data = sys.stdin.read().split()
if data:
  N = int(data[0])
  niilber = 0
  i = 1
  while i <= N and i < len(data):
    niilber = niilber + int(data[i])
    i = i + 1
  print(niilber)
```

```python
#loop03.
N = int(input())

i = 1
square = i * i

res = []
while square <= N:
  res.append(str(square))
  i += 1
  square = i * i

print(" ".join(res))
```

```python
#loop05
import sys

# Оролтыг унших
data = sys.stdin.read().split()

if data:
  N = int(data[0])

  res = []
  val = 1

  # N-ээс хэтрэхгүй 2-ын зэргүүдийг олох
  while val <= N:
    res.append(str(val))
    val = val * 2

  # Нийт хариуг зайгаар салгаж хэвлэх
  print(" ".join(res))
```

```python
#loop06
import sys

# Оролтыг унших
data = sys.stdin.read().split()

if data:
  N = int(data[0])

  val = 1
  # val хувьсагчийг N-д хүрэх эсвэл хэтрэх хүртэл 2 дахин өсгөнө
  while val < N:
    val = val * 2

  # Хэрэв яг тэнцвэл YES, үгүй бол NO
  if val == N:
    print("YES")
  else:
    print("NO")
```

```python
#loop07.
import sys

# Бүх оролтыг уншиж жагсаалт болгох
data = sys.stdin.read().split()

if data:
  count = 0
  i = 0

  # Эхний 0 хүртэл гүйж тоолох
  while i < len(data):
    if int(data[i]) == 0:
      break
    count += 1
    i += 1

  # Зөвхөн тоогоо л хэвлэнэ
  print(count)
```

```python
#loop08
import sys

# Бүх оролтыг уншиж жагсаалт болгох
data = sys.stdin.read().split()

if data:
  count = 0
  zero_count = 0
  i = 0

  # Хоёр дахь 0 хүртэл гүйж тоолох
  while i < len(data):
    val = int(data[i])
    if val == 0:
      zero_count += 1
      if zero_count == 2:
        break
    count += 1
    i += 1

  # Хоёр дахь 0 хүртэлх нийт элементүүдийн тоог хэвлэнэ
  print(count)
```

```python
#loop09
import sys

# Бүх оролтыг уншиж жагсаалт болгох
data = sys.stdin.read().split()

if data:
  max_val = int(data[0])
  i = 0

  # 0 хүртэл гүйж хамгийн их элементийг олох
  while i < len(data):
    val = int(data[i])
    if val == 0:
      break
    if val > max_val:
      max_val = val
    i += 1

  # Хамгийн их элементийг хэвлэх
  print(max_val)
```

```python
#loop10
import sys

# Бүх оролтыг уншиж жагсаалт болгох
data = sys.stdin.read().split()

if data:
  # 1. Эхлээд хамгийн их элементийг олох
  max_val = -1
  i = 0
  while i < len(data):
    val = int(data[i])
    if val == 0:
      break
    if val > max_val:
      max_val = val
    i += 1

  # 2. Хамгийн их элемент хэдэн удаа гарч ирснийг тоолох
  count = 0
  i = 0
  while i < len(data):
    val = int(data[i])
    if val == 0:
      break
    if val == max_val:
      count += 1
    i += 1

  # Үр дүнг хэвлэх
  print(count)
```
