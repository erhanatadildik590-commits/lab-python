```python
#2. Банкны дансны класс (BankAccount)
#Данс нээх, мөнгө хийх, авах, хаах, үлдэгдэл шалгах үйлдлүүд.
class BankAccount:

  def __init__(self, name, account_number, balance=0.0):
    self.name = name
    self.account_number = account_number
    self.balance = balance
    self.is_active = True

  def deposit(self, amount):
    if self.is_active:
      self.balance += amount
      print(f"{amount}₮ амжилттай нэмэгдлээ. Үлдэгдэл: {self.balance}₮")
    else:
      print("Данс хаалттай байна!")

  def withdraw(self, amount):
    if not self.is_active:
      print("Данс хаалттай байна!")
      return
    if amount <= self.balance:
      self.balance -= amount
      print(f"{amount}₮ гарлаа. Үлдэгдэл: {self.balance}₮")
    else:
      print("Дансны үлдэгдэл хүрэлцэхгүй байна!")

  def close_account(self):
    self.is_active = False
    print(f"Данс ({self.account_number}) амжилттай хаагдлаа.")

  def check_balance(self):
    if self.is_active:
      print(f"Дансны үлдэгдэл: {self.balance}₮")
    else:
      print("Данс хаалттай байна.")



acc = BankAccount("Бат", "B1001", 50000)
acc.check_balance()
acc.deposit(20000)
acc.withdraw(15000)
acc.close_account()
acc.check_balance()
```

```python
#1. Барааны класс (Product)
#N ширхэг барааны мэдээллийг жагсаалтад (list) хадгалж, хэвлэх програм.
class Product:

  def __init__(self, name, category, unit, price):
    self.name = name
    self.category = category
    self.unit = unit
    self.price = price

  def input_data(self):
    self.name = input("Барааны нэр: ")
    self.category = input("Төрөл: ")
    self.unit = input("Хэмжих нэгж: ")
    self.price = float(input("Үнэ: "))

  def print_data(self):
    print(
        f"Нэр: {self.name}, Төрөл: {self.category}, Нэгж: {self.unit}, Үнэ:"
        f" {self.price}"
    )



products = []
n = int(input("Хэдэн бараа оруулах вэ? "))
for i in range(n):
  print(f"\n{i+1}-р барааны мэдээлэл:")
  p = Product("", "", "", 0.0)
  p.input_data()
  products.append(p)

print("\n--- Бүх барааны жагсаалт ---")
for p in products:
  p.print_data()
```

```python
#3. Цаг хугацааны класс (Time)
#hours, minutes, seconds гишүүнтэй, хоёр цагийн нийт зөрүүг олох эсвэл нэмэх, форматлах класс.
class Time:

  def __init__(self, h=0, m=0, s=0):
    self.hours = h
    self.minutes = m
    self.seconds = s

  def orolt(self):
    self.hours = int(input("Цаг (hours): "))
    self.minutes = int(input("Минут (minutes): "))
    self.seconds = int(input("Секунд (seconds): "))

  def garalt(self):
    print(
        f"{self.hours:02d}:{self.minutes:02d}:{self.seconds:02d}"
    )

  def total_seconds(self):
    return self.hours * 3600 + self.minutes * 60 + self.seconds

  def difference(self, other):
    diff_sec = abs(self.total_seconds() - other.total_seconds())
    h = diff_sec // 3600
    m = (diff_sec % 3600) // 60
    s = diff_sec % 60
    return Time(h, m, s)



t1 = Time(10, 30, 0)
t2 = Time(8, 15, 45)
print("Цаг 1: ", end="")
t1.garalt()
print("Цаг 2: ", end="")
t2.garalt()

diff = t1.difference(t2)
print("Хоорондох зай (зөрүү): ", end="")
diff.garalt()
```

```python
#4-9. Комплекс тооны класс болон оператор дахин тодорхойлох (ComplexNumber)
#Бодлого 4-өөс 9 хүртэлх комплекс тооны хасах, нэмэх, хуваах, үржих болон харьцуулах (==, !=) операторуудыг нэг класс дотор бүрэн хэрэгжүүлэв.
class ComplexNumber:

  def __init__(self, real, imag):
    self.real = real
    self.imag = imag

  def __str__(self):
    if self.imag >= 0:
      return f"{self.real} + {self.imag}i"
    else:
      return f"{self.real} - {abs(self.imag)}i"


  def __sub__(self, other):
    return ComplexNumber(self.real - other.real, self.imag - other.imag)


  def __add__(self, other):
    return ComplexNumber(self.real + other.real, self.imag + other.imag)


  def __mul__(self, other):
    r = self.real * other.real - self.imag * other.imag
    i = self.real * other.imag + self.imag * other.real
    return ComplexNumber(r, i)


  def __truediv__(self, other):
    denom = other.real**2 + other.imag**2
    r = (self.real * other.real + self.imag * other.imag) / denom
    i = (self.imag * other.real - self.real * other.imag) / denom
    return ComplexNumber(r, i)


  def __eq__(self, other):
    return self.real == other.real and self.imag == other.imag


  def __ne__(self, other):
    return not (self == other)



c1 = ComplexNumber(4, 2)
c2 = ComplexNumber(1, -1)

print(f"c1 = {c1}")
print(f"c2 = {c2}")
print(f"Нэмэх (+): {c1 + c2}")
print(f"Хасах (-): {c1 - c2}")
print(f"Үржих (*): {c1 * c2}")
print(f"Хуваах (/): {c1 / c2}")
print(f"c1 == c2: {c1 == c2}")
print(f"c1 != c2: {c1 != c2}")
```

```python
#10. Энгийн ба нийлмэл удамшлын жишээ (Inheritance Example)
#Эрүүл мэндийн байгууллага эсвэл сургуулийн систем дээр суурилсан энгийн болон олон шатлалт (нийлмэл) удамшил:

class Person:

  def __init__(self, name, age):
    self.name = name
    self.age = age

  def show_info(self):
    print(f"Нэр: {self.name}, Нас: {self.age}")



class Student(Person):

  def __init__(self, name, age, gpa):
    super().__init__(name, age)
    self.gpa = gpa

  def show_student(self):
    self.show_info()
    print(f"Голч дүнгээс: {self.gpa}")



class GraduateStudent(Student):

  def __init__(self, name, age, gpa, thesis_topic):
    super().__init__(name, age, gpa)
    self.thesis_topic = thesis_topic

  def show_graduate(self):
    self.show_student()
    print(f"Дипломын сэдэв: {self.thesis_topic}")



gs = GraduateStudent("Тэмүүлэн", 22, 3.6, "AI ба Python хэл")
gs.show_graduate()
```
